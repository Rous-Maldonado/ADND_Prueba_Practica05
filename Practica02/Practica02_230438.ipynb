12.3.- **Funciones que mandan llamar otras funciones y que regresan el resultado de unn procesamiento, en este caso utilizaremos, una gfuncion predefinida en python denominada replace que nos permite sustituir letras en una cadena de texto**.

```python
def sustituir_vocales(texto):
    texto = texto.replace("A","4")
    texto = texto.replace("a","4")

    texto = texto.replace("E","3")
    texto = texto.replace("e","3")

    texto = texto.replace("I","1")
    texto = texto.replace("i","1")

    texto = texto.replace("O","0")
    texto = texto.replace("o","0")

    texto = texto.replace("U","∪")
    texto = texto.replace("u","∪")

    return texto
```

*Llamado*

```python
print(sustituir_vocales(saludar_nombre("Rosario")))
```

18.- **Eliminar los datos de los cuartiles menores**

Con la finalidad de hacer un analisis centradoen los datos de tendencia central, es decir con mayor frecuencia eliminamos los registros que entran en el cuartil menor a 0.05 y mayor a 0.95, utilizando la creación de un nuevo dataframe(df3), recordando que la muestra completa sigue en el df original(df)

```python
a=df['MILES'].quantile(0.95)
b=df['MILES'].quantile(0.05)
df3=df[df['MILES']>b]
df3=df[df['MILES']<a]
df3
```

En este código se usan los percentiles 5 y 95 de la columna MILES para encontrar valores muy bajos y muy altos dentro de los datos. La idea es hacer una limpieza de datos y evitar que valores extremos afecten el análisis. Primero se obtiene el límite inferior (b) y el límite superior (a), después se intenta filtrar el DataFrame para quedarse solo con los registros que estén dentro de ese rango. Aunque en el código el segundo filtro reemplaza al primero, la intención principal es conservar los datos más normales y eliminar posibles outliers o valores atípicos.

19.- **Cálculo de la Desviacion Estándar**

Para entender que tan varibales o dispersas son las distancias de los viajes.

```python
df3['STD']=df.MILES.std()
df3
```

20.- **Estimar el Valor de Z-Score o Puntuación Z**

con la finalidad de estimar la disperción de los datos, es decir que tan variados o diferentes son, utilizamos esta formula estadistica para poder continuar con el analisis EDA(Exploratory Data Analisys)

```python
df3['Z']=(df.MILES-df.MILES.mean())/df.MILES.std()
df3
```

21.- **Filtramos aquello que no estén en el rango de -1 y 1**

23.- **Grafico de Clasificación de Viajes por CATEGORIA**

```python
fig, ax = mat.subplots(figsize=(8,5))

grafica = sea.countplot(
    data=df,
    x='CATEGORY',
    hue='CATEGORY',
    palette='Set2',
    legend=False,
    ax=ax

 )

#Titulo
ax.set_title('Viaje por Categoría')

#Agregar etiquetas con totales
for container in ax.containers:
    ax.bar_label(container)

mat.show()
```

24.- **Grafica de viajes por Propósito(PURPOSE)**

25.- **Grafico de Lugar de Abordaje(START)**
